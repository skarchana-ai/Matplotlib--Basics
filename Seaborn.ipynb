```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df=sns.load_dataset("tips")
```

```python
df
```

```python
dataset=sns.load_dataset("diamonds")
```

```python
dataset
```

Scatter Plot

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip"
)
plt.show()
```

```python
sns.scatterplot(
    data=dataset,
    x="carat",
    y="price"
)
plt.show()
```

hue

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time"
)

plt.show()
```

```python
sns.scatterplot(
    data=dataset,
    x="carat",
    y="price",
    hue="cut"
)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="day"
)

plt.show()
```

style

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="day",
    style="sex"
)

plt.show()
```
