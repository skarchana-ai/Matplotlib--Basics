Seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df=sns.load_dataset("tips")
```

```python
df
```

```python
dataset=sns.load_dataset("diamonds")
```

```python
dataset
```

Scatter Plot

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip"
)
plt.show()
```

```python
sns.scatterplot(
    data=dataset,
    x="carat",
    y="price"
)
plt.show()
```

hue

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time"
)

plt.show()
```

```python
sns.scatterplot(
    data=dataset,
    x="carat",
    y="price",
    hue="cut"
)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="day"
)

plt.show()
```

style

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="day",
    style="sex"
)

plt.show()
```

Line Plot

```python
sns.lineplot(
    data=df,
    x="size",
    y="tip"
)

plt.show()
```

```python
sns.lineplot(
    data=dataset,
    x="carat",
    y="price"
)

plt.show()
```

Bar Plot

```python
sns.barplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
sns.barplot(
    data=dataset,
    x="depth",
    y="price"
)

plt.show()
```

Count Plot

```python
sns.countplot(
    data=df,
    x="day"
)

plt.show()
```

```python
sns.countplot(
    data=dataset,
    x="cut"
)

plt.show()
```

Box Plot (also called a box-and-whisker plot)

```python
sns.boxplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
sns.boxplot(
    data=dataset,
    x="cut",
    y="price"
)

plt.show()
```

Violin Plot

```python
sns.violinplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
sns.violinplot(
    data=dataset,
    x="cut",
    y="price"
)

plt.show()
```

Histogram

```python
sns.histplot(
    data=df,
    x="total_bill"
)

plt.show()
```

```python
sns.histplot(
    data=dataset,
    x="carat"

)
plt.show()
```

KDE

```python
sns.kdeplot(
    data=df,
    x="total_bill"
)
plt.show()
```

```python
sns.kdeplot(
    data=dataset,
    x="carat"
)

plt.show()
```

Histogram + KDE

```python
sns.histplot(
    data=df,
    x="total_bill",
    kde=True
)

plt.show()
```

```python
sns.histplot(
    data=dataset,
    x="carat",
    kde=True
)

plt.show()
```
