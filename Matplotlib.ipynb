Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]
```

```python
plt.plot(x,y)
plt.show()
```

Line Plot

```python
months =[1,2,3,4,5]
sales=[100,200,300,400,500]
```

```python
plt.plot(months,sales)
```

```python
plt.plot(months,sales)
plt.title("Sales Growth")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

Markers

```python
plt.plot(months,sales)

plt.title("Sales Growth")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.plot(months,sales,marker="o")
plt.show()
```

```python
plt.plot(months,sales)

plt.title("Sales Growth")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.plot(months, sales, marker="s")
```
